```python
# 只保留数值型列
df_numeric = df_filtered.select_dtypes(include=["number"])

# 计算相关性矩阵
correlation_matrix = df_numeric.corr()

# 提取与 RC骨龄 相关的特征
rc_correlation = correlation_matrix[["RC骨龄"]].sort_values(by="RC骨龄", ascending=False)

# 绘制热力图
plt.figure(figsize=(8, 10))  # 适当调整图像大小
sns.heatmap(rc_correlation, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
# 调整图表的边距，确保标签显示完整
plt.tight_layout()
# 保存图像
plt.savefig('rc_correlation_heatmap.png', dpi=300, bbox_inches='tight')
# 显示图像
plt.show()

# 计算RC骨龄与所有选定特征的相关性
correlation_matrix = df_filtered.corr()

# 提取RC骨龄相关性排序
rc_correlation = correlation_matrix["RC骨龄"].sort_values(ascending=False)

# 绘制相关矩阵热力图
plt.figure(figsize=(20, 16))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("特征相关性热力图")
plt.savefig('feature_correlation_heatmap.png')
plt.show()

```

```python
# 选取与 RC骨龄 相关性 ≥ 0.4 的特征
selected_features = correlation_matrix["RC骨龄"][abs(correlation_matrix["RC骨龄"]) >= 0.4].index.tolist()
selected_features.remove("RC骨龄")  # 确保移除 RC骨龄
# 提取这些特征的数据，并去除缺失值
df_selected = df[selected_features+["RC骨龄"]].dropna()
print(f"选取的特征数量: {len(selected_features)}，有效样本数: {df_selected.shape[0]}")


# **对比清洗前后的分布**
plt.figure(figsize=(8, 5))
sns.histplot(df["RC骨龄"], bins=30, kde=True, color="gray", alpha=0.6, label="清洗前")
sns.histplot(df_selected["RC骨龄"], bins=30, kde=True, color="blue", alpha=0.7, label="清洗后")
plt.xlabel("RC骨龄")
plt.ylabel("频数")
plt.title("RC骨龄 清洗前后分布对比")
plt.legend()
plt.savefig('rc_distribution_clean_before_and_after.png')
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import numpy as np



# **数据标准化（不包含 `RC骨龄` 和 `年龄（月）`）**
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_selected[selected_features])  # 仅基于选定特征进行标准化

# **进行 KMeans 聚类分析**
num_clusters = 8  # 设定聚类数量（可以调整）
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
df_selected["聚类类别"] = kmeans.fit_predict(X_scaled)

# **PCA 降维到 2 维进行可视化**
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# **绘制聚类散点图**
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df_selected["聚类类别"], cmap="viridis", alpha=0.7)
plt.xlabel("主成分 1")
plt.ylabel("主成分 2")
plt.title("RC骨龄 相关特征的 KMeans 聚类分析（不含 RC骨龄）")
plt.colorbar(label="聚类类别")
plt.savefig('kmeans_pca.png')
plt.show()

# **按 6 个月为单位对年龄进行分箱**
df_selected["骨龄分组"] = pd.cut(df_selected["RC骨龄"], bins=np.arange(0, df_selected["RC骨龄"].max() + 6, 6), right=False)

# **统计每个年龄分组的聚类分布**
age_cluster_counts = df_selected.groupby(["骨龄分组", "聚类类别"]).size().unstack(fill_value=0)

# **绘制年龄分组的聚类分布条形图**
age_cluster_counts.plot(kind="bar", stacked=True, figsize=(12, 6), colormap="viridis")
plt.xlabel("RC骨龄")
plt.ylabel("样本数")
plt.title("不同骨龄分组分组的 KMeans 聚类分布")
plt.legend(title="聚类类别")
plt.xticks(rotation=45)
plt.savefig('kmeans_ba_dist.png')
plt.show()

# **输出每个聚类类别的统计信息**
cluster_counts = df_selected["聚类类别"].value_counts().reset_index()
cluster_counts.columns = ["聚类类别", "样本数"]
print("\n=== 各聚类类别的样本数量 ===")
print(cluster_counts)

# **绘制不同聚类类别的 RC骨龄 分布**
plt.figure(figsize=(8, 5))
sns.boxplot(x="聚类类别", y="RC骨龄", data=df_selected)
plt.xlabel("聚类类别")
plt.ylabel("RC骨龄")
plt.title("不同聚类类别的 RC骨龄 分布")
plt.savefig('kmeans_ba_box.png')
plt.show()
```
